# Figure 4c: lateral propagation and source attenuation

Run `python campaigns/s1_s6_azimuthal/compute_attenuation_cache.py` from `refactored/` first. The default caches use `azimean_preforcing` and the symmetric lock-in reduction.


```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from dsmclib.attenuation import plot_attenuation_panels
from dsmclib.paths import output_dir
```

```python
CACHE_DIR = output_dir('s1_s6_azimuthal', 'cache')
FIGURE_DIR = output_dir('s1_s6_azimuthal', 'figures')
BACKGROUND = 'azimean_preforcing'
ALIASES = ['S4', 'S5', 'S6']

reductions = {}
for alias in ALIASES:
    path = CACHE_DIR / f'lockin_attenuation_{alias}_{BACKGROUND}.npz'
    with np.load(path, allow_pickle=False) as archive:
        reduction = {key: archive[key] for key in archive.files}
    reductions[int(reduction['n_azi'])] = reduction
```

```python
fig = plot_attenuation_panels(reductions)
output_path = FIGURE_DIR / 'figure4c_symmetric_lockin_attenuation.png'
fig.savefig(output_path, dpi=300, bbox_inches='tight')
print(output_path)
plt.show()
```
